```python
# Install necessary libraries
import google.generativeai as genai

# Configure Gemini API with your API key
genai.configure(api_key="")
model = genai.GenerativeModel("gemini-1.0-pro")
```

```python


# Link to the JSON file (can be accessed via Google Drive link)
JSON_FILE_REFERENCE = "https://github.com/AI-M-U/MetaMajor-Prompts/blob/main/courses_data.json"

# Function to generate a 2-year meta-major plan based on user input
def generate_meta_major_plan(majors):
    try:
        prompt = (   f"""As a AI assistance, generate a 2-year program based on the following majors: {', '.join(majors)} .
                        1- Title for this Meta Major
                        2- list of courses: Retrive the courses for each semester from the current courses provided here {JSON_FILE_REFERENCE}.
                        example  output course details:
                        course_id: COS 125
                        course_title: "**Introduction to Problem Solving Using Computer Programming**",
                        Do not generate your own course use current majors courses provided.
                        """
                        )

        # Request the Gemini API to generate the plan
        response = model.generate_content(prompt)
        suggestion = response.text.strip()

        return suggestion

    except Exception as e:
        return f"Error generating plan: {str(e)}"

# Input the majors as a comma-separated list
majors_input = input("Enter a list of majors (comma-separated): ")
majors_list = [major.strip() for major in majors_input.split(",")]

# Generate and display the meta-major plan
plan = generate_meta_major_plan(majors_list)

print("\nGenerated Meta-Major Plan:")
print(plan)
```
